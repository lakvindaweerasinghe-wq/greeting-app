```python
import random
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, random_split
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt


def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)


base_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2470, 0.2435, 0.2616))
])

full_train_dataset = torchvision.datasets.CIFAR10(
    root="./data", train=True, download=True, transform=base_transform
)
test_dataset = torchvision.datasets.CIFAR10(
    root="./data", train=False, download=True, transform=base_transform
)


train_size = 40000
val_size = len(full_train_dataset) - train_size
train_dataset, val_dataset = random_split(
    full_train_dataset, [train_size, val_size],
    generator=torch.Generator().manual_seed(42)
)

batch_size = 128
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
```

```python

class MLPClassifier(nn.Module):
    def __init__(self, input_dim=32 * 32 * 3, num_classes=10):
        super().__init__()

        self.model = nn.Sequential(
            nn.Flatten(),                  # (Batch, 3, 32, 32) -> (Batch, 3072)
            nn.Linear(input_dim, 512),     # Trainable Layer 1: 3072 -> 512
            nn.ReLU(),
            nn.Linear(512, 128),           # Trainable Layer 2: 512 -> 128
            nn.ReLU(),
            nn.Linear(128, num_classes)    # Trainable Layer 3: 128 -> 10
        )

    def forward(self, x):
        return self.model(x)



set_seed(42)
mlp_model = MLPClassifier().to(device)


mlp_params = sum(p.numel() for p in mlp_model.parameters() if p.requires_grad)
print(f"MLP Trainable Parameters: {mlp_params:,}")


criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(mlp_model.parameters(), lr=0.001)



def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    epoch_loss = running_loss / total
    epoch_acc = 100.0 * correct / total
    return epoch_loss, epoch_acc


def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)

            running_loss += loss.item() * images.size(0)
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    avg_loss = running_loss / total
    accuracy = 100.0 * correct / total
    return avg_loss, accuracy



num_epochs = 15
mlp_train_losses, mlp_val_losses = [], []
mlp_train_accs, mlp_val_accs = [], []

for epoch in range(1, num_epochs + 1):
    train_loss, train_acc = train_one_epoch(mlp_model, train_loader, criterion, optimizer, device)
    val_loss, val_acc = evaluate(mlp_model, val_loader, criterion, device)

    mlp_train_losses.append(train_loss)
    mlp_val_losses.append(val_loss)
    mlp_train_accs.append(train_acc)
    mlp_val_accs.append(val_acc)

    print(
        f"Epoch [{epoch:02d}/{num_epochs:02d}] | "
        f"Train Loss: {train_loss:.4f}, Train Acc: {train_acc:.2f}% | "
        f"Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.2f}%"
    )


final_train_loss, final_train_acc = evaluate(mlp_model, train_loader, criterion, device)
test_loss, test_acc = evaluate(mlp_model, test_loader, criterion, device)

print("\n--- Exercise 1 Final Summary (MLP) ---")
print(f"Final Train Accuracy : {final_train_acc:.2f}% (Loss: {final_train_loss:.4f})")
print(f"Test Accuracy        : {test_acc:.2f}% (Loss: {test_loss:.4f})")


plt.figure(figsize=(8, 5))
plt.plot(range(1, num_epochs + 1), mlp_train_losses, marker="o", label="Training Loss")
plt.plot(range(1, num_epochs + 1), mlp_val_losses, marker="s", label="Validation Loss")
plt.title("Exercise 1: MLP Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss (CrossEntropy)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
class CNNClassifier(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()

        self.conv1 = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, padding=1)
        self.relu = nn.ReLU()
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)


        self.conv2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, padding=1)


        self.fc = nn.Linear(64 * 8 * 8, num_classes)

    def forward(self, x):
        x = self.pool(self.relu(self.conv1(x)))
        x = self.pool(self.relu(self.conv2(x)))
        x = x.view(x.size(0), -1)
        x = self.fc(x)
        return x



set_seed(42)
cnn_model = CNNClassifier().to(device)

cnn_params = sum(p.numel() for p in cnn_model.parameters() if p.requires_grad)
print(f"MLP Trainable Parameters : {mlp_params:,}")
print(f"CNN Trainable Parameters : {cnn_params:,}")

cnn_criterion = nn.CrossEntropyLoss()
cnn_optimizer = optim.Adam(cnn_model.parameters(), lr=0.001)

cnn_train_losses, cnn_val_losses = [], []
cnn_train_accs, cnn_val_accs = [], []

for epoch in range(1, num_epochs + 1):
    train_loss, train_acc = train_one_epoch(cnn_model, train_loader, cnn_criterion, cnn_optimizer, device)
    val_loss, val_acc = evaluate(cnn_model, val_loader, cnn_criterion, device)

    cnn_train_losses.append(train_loss)
    cnn_val_losses.append(val_loss)
    cnn_train_accs.append(train_acc)
    cnn_val_accs.append(val_acc)

    print(
        f"Epoch [{epoch:02d}/{num_epochs:02d}] | "
        f"Train Loss: {train_loss:.4f}, Train Acc: {train_acc:.2f}% | "
        f"Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.2f}%"
    )

final_cnn_train_loss, final_cnn_train_acc = evaluate(cnn_model, train_loader, cnn_criterion, device)
cnn_test_loss, cnn_test_acc = evaluate(cnn_model, test_loader, cnn_criterion, device)

print("\n--- Exercise 2 Final Summary (CNN) ---")
print(f"Final Train Accuracy : {final_cnn_train_acc:.2f}% (Loss: {final_cnn_train_loss:.4f})")
print(f"Test Accuracy        : {cnn_test_acc:.2f}% (Loss: {cnn_test_loss:.4f})")


plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(range(1, num_epochs + 1), cnn_train_losses, marker="o", label="CNN Train Loss")
plt.plot(range(1, num_epochs + 1), cnn_val_losses, marker="s", label="CNN Val Loss")
plt.title("Exercise 2: CNN Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss (CrossEntropy)")
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(range(1, num_epochs + 1), mlp_val_losses, marker="o", linestyle="--", label="MLP Val Loss")
plt.plot(range(1, num_epochs + 1), cnn_val_losses, marker="s", label="CNN Val Loss")
plt.title("Validation Loss Comparison: MLP vs CNN")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
baseline_train_acc = final_cnn_train_acc
baseline_test_acc = cnn_test_acc
baseline_gap = baseline_train_acc - baseline_test_acc



class CNNWithDropout(nn.Module):
    def __init__(self, num_classes=10, dropout_p=0.5):
        super().__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, padding=1)
        self.relu = nn.ReLU()
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        self.conv2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, padding=1)
        self.dropout = nn.Dropout(p=dropout_p)
        self.fc = nn.Linear(64 * 8 * 8, num_classes)

    def forward(self, x):
        x = self.pool(self.relu(self.conv1(x)))
        x = self.pool(self.relu(self.conv2(x)))
        x = x.view(x.size(0), -1)
        x = self.dropout(x)
        x = self.fc(x)
        return x


set_seed(42)
dropout_model = CNNWithDropout(num_classes=10, dropout_p=0.5).to(device)
drop_criterion = nn.CrossEntropyLoss()
drop_optimizer = optim.Adam(dropout_model.parameters(), lr=0.001)

print("Training CNN + Dropout...")
for epoch in range(1, num_epochs + 1):
    t_loss, t_acc = train_one_epoch(dropout_model, train_loader, drop_criterion, drop_optimizer, device)
    v_loss, v_acc = evaluate(dropout_model, val_loader, drop_criterion, device)
    print(f"[Dropout] Epoch [{epoch:02d}/{num_epochs:02d}] | Train Acc: {t_acc:.2f}% | Val Acc: {v_acc:.2f}%")

_, drop_train_acc = evaluate(dropout_model, train_loader, drop_criterion, device)
_, drop_test_acc = evaluate(dropout_model, test_loader, drop_criterion, device)
drop_gap = drop_train_acc - drop_test_acc



aug_transform = transforms.Compose([
    transforms.RandomRotation(degrees=15),
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2470, 0.2435, 0.2616))
])


aug_full_train = torchvision.datasets.CIFAR10(
    root="./data", train=True, download=False, transform=aug_transform
)
aug_train_dataset = torch.utils.data.Subset(aug_full_train, train_dataset.indices)
aug_train_loader = DataLoader(aug_train_dataset, batch_size=batch_size, shuffle=True)

set_seed(42)
aug_model = CNNClassifier(num_classes=10).to(device)
aug_criterion = nn.CrossEntropyLoss()
aug_optimizer = optim.Adam(aug_model.parameters(), lr=0.001)

print("\nTraining Unchanged Baseline CNN + Data Augmentation...")
for epoch in range(1, num_epochs + 1):
    t_loss, t_acc = train_one_epoch(aug_model, aug_train_loader, aug_criterion, aug_optimizer, device)
    v_loss, v_acc = evaluate(aug_model, val_loader, aug_criterion, device)
    print(f"[Augmentation] Epoch [{epoch:02d}/{num_epochs:02d}] | Train Acc: {t_acc:.2f}% | Val Acc: {v_acc:.2f}%")


_, aug_train_acc = evaluate(aug_model, train_loader, aug_criterion, device)
_, aug_test_acc = evaluate(aug_model, test_loader, aug_criterion, device)
aug_gap = aug_train_acc - aug_test_acc


methods = ["Baseline CNN", "CNN + Dropout (p=0.5)", "CNN + Data Augmentation"]
train_accs = [baseline_train_acc, drop_train_acc, aug_train_acc]
test_accs = [baseline_test_acc, drop_test_acc, aug_test_acc]
gaps = [baseline_gap, drop_gap, aug_gap]

print("\n--- Exercise 3 Comparative Summary ---")
print(f"{'Method':<28} | {'Train Acc':>10} | {'Test Acc':>10} | {'Gap (Train-Test)':>16}")
print("-" * 74)
for m, tr, te, gp in zip(methods, train_accs, test_accs, gaps):
    print(f"{m:<28} | {tr:>9.2f}% | {te:>9.2f}% | {gp:>15.2f}%")



x = np.arange(len(methods))
width = 0.25

plt.figure(figsize=(9, 5))
plt.bar(x - width, train_accs, width, label="Train Accuracy (%)")
plt.bar(x, test_accs, width, label="Test Accuracy (%)")
plt.bar(x + width, gaps, width, label="Train–Test Gap (%)")

plt.title("Exercise 3: Generalisation Comparison Across Regularisation Methods")
plt.xlabel("Model / Regularisation Technique")
plt.ylabel("Percentage (%)")
plt.xticks(x, methods, rotation=10)
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
class FlexibleCNN(nn.Module):
    def __init__(self, channels_list, num_classes=10):
        super().__init__()
        layers = []
        in_c = 3
        spatial_size = 32

        for out_c in channels_list:

            if spatial_size < 2:
                break
            layers.append(nn.Conv2d(in_c, out_c, kernel_size=3, padding=1))
            layers.append(nn.ReLU())
            layers.append(nn.MaxPool2d(kernel_size=2, stride=2))
            in_c = out_c
            spatial_size //= 2

        self.features = nn.Sequential(*layers)
        self.flatten_dim = in_c * spatial_size * spatial_size
        self.fc = nn.Linear(self.flatten_dim, num_classes)

    def forward(self, x):
        x = self.features(x)
        x = x.view(x.size(0), -1)
        x = self.fc(x)
        return x


ex4_configs = [
    ("D1_W1 (1 Conv, [16])",        [16]),
    ("D1_W2 (1 Conv, [32])",        [32]),
    ("D1_W3 (1 Conv, [64])",        [64]),
    ("D2_W1 (2 Conv, [16, 32])",    [16, 32]),
    ("D2_W2 (2 Conv, [32, 64])",    [32, 64]),
    ("D2_W3 (2 Conv, [64, 128])",   [64, 128]),
]


ex4_configs = [
    ("Depth1_W1 [16]",          [16]),
    ("Depth1_W2 [32]",          [32]),
    ("Depth2_W1 [16, 32]",      [16, 32]),
    ("Depth2_W2 [32, 64]",      [32, 64]),
    ("Depth3_W1 [16, 32, 64]",  [16, 32, 64]),
    ("Depth3_W2 [32, 64, 128]", [32, 64, 128]),
]

ex4_results = []
ex4_epochs = 10

for name, ch_list in ex4_configs:
    set_seed(42)
    model = FlexibleCNN(ch_list, num_classes=10).to(device)
    param_count = sum(p.numel() for p in model.parameters() if p.requires_grad)

    opt = optim.Adam(model.parameters(), lr=0.001)
    crit = nn.CrossEntropyLoss()

    last_train_loss = None
    last_val_acc = None

    for epoch in range(1, ex4_epochs + 1):
        tr_loss, tr_acc = train_one_epoch(model, train_loader, crit, opt, device)
        val_loss, val_acc = evaluate(model, val_loader, crit, device)
        last_train_loss = tr_loss
        last_val_acc = val_acc

    ex4_results.append({
        "name": name,
        "params": param_count,
        "train_loss": last_train_loss,
        "val_acc": last_val_acc
    })
    print(f"{name:<26} | Params: {param_count:>9,} | Train Loss: {last_train_loss:.4f} | Val Acc: {last_val_acc:.2f}%")

print("\n--- Exercise 4 Summary Table ---")
print(f"{'Configuration':<26} | {'Trainable Params':>16} | {'Train Loss':>11} | {'Val Acc':>9}")
print("-" * 72)
for r in ex4_results:
    print(f"{r['name']:<26} | {r['params']:>16,} | {r['train_loss']:>11.4f} | {r['val_acc']:>8.2f}%")

names = [r["name"] for r in ex4_results]
val_accs = [r["val_acc"] for r in ex4_results]
params_list = [r["params"] for r in ex4_results]

fig, ax1 = plt.subplots(figsize=(10, 5))

bars = ax1.bar(names, val_accs, color="steelblue", alpha=0.85, label="Validation Accuracy (%)")
ax1.set_xlabel("Model Depth & Width Configuration")
ax1.set_ylabel("Validation Accuracy (%)", color="steelblue")
ax1.tick_params(axis="y", labelcolor="steelblue")
plt.xticks(rotation=20, ha="right")
ax1.grid(axis="y", linestyle="--", alpha=0.5)

ax2 = ax1.twinx()
ax2.plot(names, params_list, color="crimson", marker="o", linewidth=2, label="Trainable Parameters")
ax2.set_ylabel("Trainable Parameters", color="crimson")
ax2.tick_params(axis="y", labelcolor="crimson")

plt.title("Exercise 4: Validation Accuracy & Parameter Count Across Width/Depth Variations")
fig.tight_layout()
plt.show()
```
